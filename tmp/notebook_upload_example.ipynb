```python
# unzip and list content programmatically
import zipfile
from pathlib import Path
z = Path('tmp/pyragify_cncdocs.zip')
if z.exists():
    with zipfile.ZipFile(z) as zf:
        print('Files in zip:')
        for name in zf.namelist():
            print(' -', name)
else:
    print('Zip not found -- run ./scripts/generate-notebooklm.sh first')
```
